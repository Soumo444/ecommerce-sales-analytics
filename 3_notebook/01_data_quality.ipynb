```python
# Step 3: Preview the data
print("--- Data Head ---")
display(df.head(3))

# Step 4: Missing values check
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

print("\n--- Missing Values Report ---")
display(missing_df)
```

```python
# Step 5: Duplicate check
print("Total duplicate rows:", df.duplicated().sum())
print("Duplicate Transaction IDs (TID):", df["TID"].duplicated().sum())

# Step 6: Unique values count
print("\n--- Unique Values in Each Column ---")
print(df.nunique())
```

```python
# Step 7: Check categorical values distribution
categorical_columns = [
    "Gender",
    "Age Group",
    "Product Category",
    "Discount Availed",
    "Discount Name",
    "Purchase Method",
    "Location"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))

# Step 8: Numerical summary statistics
print("\n--- Numerical Summary ---")
display(df.describe().T)
```

```python
# Step 9: Check financial logic (Gross - Discount = Net)
df["Calculated Net Amount"] = df["Gross Amount"] - df["Discount Amount (INR)"]
df["Amount Difference"] = abs(df["Calculated Net Amount"] - df["Net Amount"])

incorrect_amounts = df[df["Amount Difference"] > 0.01]
print("Rows with financial calculation issues:", len(incorrect_amounts))

# Step 10: Check impossible negative values
numeric_columns = ["Discount Amount (INR)", "Gross Amount", "Net Amount"]
for column in numeric_columns:
    print(column, "negative values:", (df[column] < 0).sum())

# Generate Data Quality Report Summary Table
quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique()
})

print("\n--- DATA QUALITY REPORT ---")
display(quality_report)
```

```python
import pandas as pd
import numpy as np
import os

# 1. Load raw data
df = pd.read_csv(r'D:\e-commerce-data-analytics-project\1_Data\1_raw\project1_df.csv')

print(f"Original row count: {len(df)}")

# 2. Clean negative Net Amounts (Filter out rows where Net Amount < 0)
df_cleaned = df[df['Net Amount'] >= 0].copy()

print(f"Row count after removing negative net amounts: {len(df_cleaned)}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")

# 3. Fill missing Discount Names with 'No Discount' for clarity
df_cleaned['Discount Name'] = df_cleaned['Discount Name'].fillna('No Discount')

# 4. Extract useful date features for time-series analysis later
df_cleaned['Purchase Date'] = pd.to_datetime(df_cleaned['Purchase Date'])
df_cleaned['Year'] = df_cleaned['Purchase Date'].dt.year
df_cleaned['Month'] = df_cleaned['Purchase Date'].dt.month
df_cleaned['Month Name'] = df_cleaned['Purchase Date'].dt.strftime('%B')
df_cleaned['Day of Week'] = df_cleaned['Purchase Date'].dt.day_name()

# 5. Ensure the target directory exists, then save the cleaned dataset
output_dir = r'D:\e-commerce-data-analytics-project\1_Data\1_processed'
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(output_dir, 'ecommerce_cleaned.csv')
df_cleaned.to_csv(output_path, index=False)

print(f"\nSuccess! Cleaned dataset saved to: {output_path}")
```
