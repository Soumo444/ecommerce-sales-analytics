```python
import os
import re
import sqlite3
import pandas as pd

# 1. Connect to SQLite database (in-memory or file-based)
conn = sqlite3.connect(":memory:")  # Or use sqlite3.connect("ecommerce.db")

# 2. Load cleaned CSV into the 'ecommerce' SQLite table
csv_path = r"D:\e-commerce-data-analytics-project\1_Data\2_processed\ecommerce_cleaned.csv"
df_raw = pd.read_csv(csv_path)

# Write DataFrame to SQLite table named 'ecommerce'
df_raw.to_sql("ecommerce", conn, if_exists="replace", index=False)

# 3. Read the SQL file containing queries
sql_file_path = r"D:\e-commerce-data-analytics-project\5_SQL Business Analysi\02_business_insights.sql"
with open(sql_file_path, "r") as file:
    sql_script = file.read()

# 4. Split script into individual queries
raw_queries = [q.strip() for q in sql_script.split(";") if q.strip()]

# 5. Execute each query with its actual title header
for query in raw_queries:
    # Extract the comment line (e.g. "-- 1. EXECUTIVE REVENUE SUMMARY") as title
    match = re.search(r"--\s*(.+)", query)
    title = match.group(1).strip() if match else "SQL Query Result"

    print("==================================================")
    print(f"📊 {title}")
    print("==================================================")

    df_result = pd.read_sql_query(query, conn)
    display(df_result)
    print("\n")
```

```python
import re
import sqlite3
import pandas as pd

# 1. Create an in-memory database connection
conn = sqlite3.connect(":memory:")

# 2. Load cleaned CSV dataset into SQLite
csv_path = r"D:\e-commerce-data-analytics-project\1_Data\2_processed\ecommerce_cleaned.csv"
df_raw = pd.read_csv(csv_path)
df_raw.to_sql("ecommerce", conn, if_exists="replace", index=False)

# 3. Read 04_discount_analysis.sql file
sql_file_path = r"D:\e-commerce-data-analytics-project\5_SQL Business Analysi\04_discount_analysis.sql"
with open(sql_file_path, "r") as file:
    sql_script = file.read()

# 4. Split into individual queries by semicolon
raw_queries = [q.strip() for q in sql_script.split(";") if q.strip()]

# 5. Execute each query with clean title extraction
for query in raw_queries:
    # Filter out banner lines (===) and extract actual text title comments
    titles = [
        line.replace("--", "").strip()
        for line in query.splitlines()
        if line.strip().startswith("--") and "=" not in line
    ]

    title = titles[0] if titles else "DISCOUNT ANALYSIS QUERY"

    print("==================================================")
    print(f"📊 {title}")
    print("==================================================")

    # Run query and display output
    df_result = pd.read_sql_query(query, conn)
    display(df_result)
    print("\n")

# 6. Close connection safely
conn.close()
```
